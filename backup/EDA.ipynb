```python
import pandas as pd
import yfinance as yf
import requests
from io import StringIO



import pandas as pd
import requests
from io import StringIO

def get_sp500_tickers():
    url = 'https://en.wikipedia.org/wiki/List_of_S%26P_500_companies'
    
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
    }
    
    response = requests.get(url, headers=headers)
    response.raise_for_status() 
    
    # Use the 'attrs' argument to specifically find the table with id="constituents"
    tables = pd.read_html(StringIO(response.text), attrs={'id': 'constituents'})
    
    # Because we searched by a unique ID, our target table will be the first and only item in the list
    df = tables[0]
    
    # Extract and clean the ticker symbols
    tickers = df['Symbol'].tolist()
    cleaned_tickers = [ticker.replace('.', '-') for ticker in tickers]
    
    return cleaned_tickers

# Test it out
tickers = get_sp500_tickers()
print(f"Successfully scraped {len(tickers)} tickers from the 'constituents' table.")

# sp500_tickers = pd.read_csv('SP500.csv')['Symbol'].tolist()
# print(f"Found {len(sp500_tickers)} tickers.")

# 2. Download the data
print("Downloading historical data from Yahoo Finance...")

data = yf.download(
    tickers = tickers,
    start = "2015-01-01", 
    end = "2025-01-01",
    group_by = 'ticker', 
    auto_adjust = True,  
    threads = True       
)

print("Download complete!")
```

```python
# Group tickers by sector and calculate the average daily return for each sector
sector_returns = pd.DataFrame()
for sector in SP_df['Sector'].unique():
    sector_tickers = SP_df[SP_df['Sector'] == sector]['Symbol'].tolist()
    valid_tickers = [t for t in sector_tickers if t in returns.columns]
    if valid_tickers:
        sector_returns[sector] = returns[valid_tickers].mean(axis=1)

# Calculate cumulative returns
cumulative_sector_returns = (1 + sector_returns).cumprod()

# Plot cumulative returns by sector
plt.figure(figsize=(14, 8))
for sector in cumulative_sector_returns.columns:
    plt.plot(cumulative_sector_returns.index, cumulative_sector_returns[sector], label=sector)
plt.title('Cumulative Returns by Sector (2015-2025)')
plt.xlabel('Date')
plt.ylabel('Cumulative Return')
plt.legend(loc='upper left', bbox_to_anchor=(1, 1))
plt.grid(True)
plt.show()
```

## 2. Maximum Drawdown Analysis
Identifying the largest peak-to-trough drops for each sector.


```python
def calculate_max_drawdown(return_series):
    comp_ret = (1 + return_series).cumprod()
    peak = comp_ret.expanding(min_periods=1).max()
    dd = (comp_ret / peak) - 1
    return dd.min()

max_drawdowns = sector_returns.apply(calculate_max_drawdown)
plt.figure(figsize=(12, 6))
max_drawdowns.sort_values().plot(kind='bar', color='salmon')
plt.title('Maximum Drawdown by Sector (2015-2025)')
plt.ylabel('Max Drawdown')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.show()

```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# Step 1: Data Cleaning (Based on your heatmap findings)
# ---------------------------------------------------------
# Drop tickers that are missing more than 10% of their data
threshold = int(0.10 * len(close_data))
clean_data = close_data.dropna(thresh=len(close_data) - threshold, axis=1)

# Forward-fill remaining missing data, then backward-fill any leading NaNs
clean_data = clean_data.ffill().bfill()

print(f"Original shape: {close_data.shape}")
print(f"Cleaned shape: {clean_data.shape}")

# Calculate daily returns
returns = clean_data.pct_change().dropna()

# ---------------------------------------------------------
# Step 2: Calculate Heuristic Information (eta)
# ---------------------------------------------------------
# We will use the individual annualized Sharpe Ratio as the heuristic
# to guide the ants toward fundamentally good assets.
annualized_returns = returns.mean() * 252
annualized_volatility = returns.std() * np.sqrt(252)
individual_sharpe = annualized_returns / annualized_volatility

# ACO probabilities must be positive. We shift the heuristic values so the minimum is slightly above 0.
heuristic_info = individual_sharpe - individual_sharpe.min() + 0.01

# ---------------------------------------------------------
# Step 3: Ant Colony Optimization Setup
# ---------------------------------------------------------
tickers = clean_data.columns.tolist()
num_assets = len(tickers)
target_assets = 30

# ACO Hyperparameters
num_ants = 100           # Number of portfolios generated per iteration
num_iterations = 500     # Number of search cycles
alpha = 1.0             # Importance of pheromones (historical algorithm success)
beta = 2.0              # Importance of heuristic (Sharpe Ratio)
evaporation_rate = 0.1  # Rate at which old pheromones decay
Q = 1.0                 # Pheromone deposit multiplier

# Initialize pheromones equally across all assets
pheromones = np.ones(num_assets)

# Tracking the best overall portfolio
global_best_portfolio = None
global_best_fitness = -np.inf
convergence_curve = []

# ---------------------------------------------------------
# Step 4: Run the ACO Algorithm
# ---------------------------------------------------------
for iteration in range(num_iterations):
    ant_portfolios = []
    ant_fitnesses = []
    ant_portfolios_sharpe = []
    for ant in range(num_ants):
        # Calculate selection probabilities: (Pheromone^alpha) * (Heuristic^beta)
        probabilities = (pheromones ** alpha) * (heuristic_info.values ** beta)
        probabilities /= probabilities.sum() # Normalize to sum to 1
        
        # Ant selects exactly 30 unique assets based on the probabilities
        selected_indices = np.random.choice(
            range(num_assets), 
            size=target_assets, 
            replace=False, 
            p=probabilities
        )
        
        # Evaluate Fitness: Calculate the Sharpe Ratio of these 30 assets (Equally Weighted)
        selected_returns = returns.iloc[:, selected_indices]
        portfolio_daily_return = selected_returns.mean(axis=1) 
        
        port_ann_return = portfolio_daily_return.mean() * 252
        port_ann_vol = portfolio_daily_return.std() * np.sqrt(252)
        
        # Prevent division by zero
        portfolio_sharpe = port_ann_return / port_ann_vol if port_ann_vol > 0 else 0

        fitnesses = (0.5 * port_ann_vol) + (0.5 * port_ann_return)
        ant_portfolios.append(selected_indices)
        ant_fitnesses.append(portfolio_sharpe)
        ant_portfolios_sharpe.append(portfolio_sharpe)
        
    # Find the best ant in this specific iteration
    iter_best_idx = np.argmax(ant_fitnesses)
    iter_best_portfolio = ant_portfolios[iter_best_idx]
    iter_best_fitness = ant_fitnesses[iter_best_idx]
    iter_best_sharpe = ant_portfolios_sharpe[iter_best_idx]
    # Update global best if this iteration found a better portfolio
    if iter_best_fitness > global_best_fitness:
        global_best_fitness = iter_best_fitness
        global_best_portfolio = iter_best_portfolio
        global_best_sharpe = iter_best_sharpe
        
    convergence_curve.append(global_best_sharpe)
    
    # Pheromone Update Phase
    # 1. Evaporation (Apply to all assets)
    pheromones *= (1 - evaporation_rate)
    
    # 2. Deposit (Reward the assets chosen by the best ant of the iteration)
    for idx in iter_best_portfolio:
        pheromones[idx] += Q * iter_best_fitness

# ---------------------------------------------------------
# Step 5: Output the Results
# ---------------------------------------------------------
best_tickers = [tickers[i] for i in global_best_portfolio]

print("\n--- ACO Optimization Complete ---")
print(f"Best Portfolio Sharpe Ratio: {global_best_fitness:.4f}")
print("\nTop 30 Selected Assets:")
for i, ticker in enumerate(best_tickers, 1):
    print(f"{i}. {ticker}")

# Plot the convergence
plt.figure(figsize=(10, 5))
plt.plot(convergence_curve, marker='o', color='b')
plt.title('ACO Convergence: Best Portfolio Sharpe Ratio over Iterations')
plt.xlabel('Iteration')
plt.ylabel('Sharpe Ratio')
plt.grid(True)
plt.show()
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# Step 1: Data Cleaning (Based on your heatmap findings)
# ---------------------------------------------------------
# Drop tickers that are missing more than 10% of their data
threshold = int(0.10 * len(close_data))
clean_data = close_data.dropna(thresh=len(close_data) - threshold, axis=1)

# Forward-fill remaining missing data, then backward-fill any leading NaNs
clean_data = clean_data.ffill().bfill()

print(f"Original shape: {close_data.shape}")
print(f"Cleaned shape: {clean_data.shape}")

# Calculate daily returns
returns = clean_data.pct_change().dropna()

# ---------------------------------------------------------
# Step 2: Calculate Heuristic Information (eta)
# ---------------------------------------------------------
# For Maximum Decorrelation, we want our heuristic to guide ants 
# toward assets that have low correlation with the rest of the market.
tickers = clean_data.columns.tolist()
num_assets = len(tickers)

# Calculate the full correlation matrix
full_corr_matrix = returns.corr()

# Calculate each asset's average correlation with all OTHER assets
avg_asset_corr = (full_corr_matrix.sum(axis=0) - 1) / (num_assets - 1)

# Inverse the correlation so that lower correlation = higher heuristic score
# Correlation ranges from [-1, 1], so (1 - corr) ranges from [0, 2]
heuristic_info = 1.0 - avg_asset_corr

# Shift to ensure all probabilities are strictly > 0
heuristic_info = heuristic_info - heuristic_info.min() + 0.01

# ---------------------------------------------------------
# Step 3: Ant Colony Optimization Setup
# ---------------------------------------------------------
target_assets = 30

# ACO Hyperparameters
num_ants = 100           # Number of portfolios generated per iteration
num_iterations = 500     # Number of search cycles
alpha = 1.0              # Importance of pheromones (historical algorithm success)
beta = 2.0               # Importance of heuristic (Low Market Correlation)
evaporation_rate = 0.1   # Rate at which old pheromones decay
Q = 1.0                  # Pheromone deposit multiplier

# Initialize pheromones equally across all assets
pheromones = np.ones(num_assets)

# Tracking the best overall portfolio
global_best_portfolio = None
global_best_fitness = -np.inf
global_best_sharpe = 0   # We can still track Sharpe just to see it
convergence_curve = []

# ---------------------------------------------------------
# Step 4: Run the ACO Algorithm
# ---------------------------------------------------------
for iteration in range(num_iterations):
    ant_portfolios = []
    ant_fitnesses = []
    ant_portfolios_sharpe = []
    
    for ant in range(num_ants):
        # Calculate selection probabilities: (Pheromone^alpha) * (Heuristic^beta)
        probabilities = (pheromones ** alpha) * (heuristic_info.values ** beta)
        probabilities /= probabilities.sum() # Normalize to sum to 1
        
        # Ant selects exactly 30 unique assets based on the probabilities
        selected_indices = np.random.choice(
            range(num_assets), 
            size=target_assets, 
            replace=False, 
            p=probabilities
        )
        
        selected_returns = returns.iloc[:, selected_indices]
        
        # --- NEW FITNESS: Maximum Decorrelation ---
        # 1. Calculate the correlation matrix of the selected 30 assets
        sel_corr_matrix = selected_returns.corr().values
        
        # 2. Extract the upper triangle (off-diagonal pairwise correlations)
        row_idx, col_idx = np.triu_indices(target_assets, k=1)
        avg_pairwise_corr = sel_corr_matrix[row_idx, col_idx].mean()
        
        # 3. Fitness function: We want to MINIMIZE correlation.
        # By doing 1.0 - avg_pairwise_corr, a lower correlation results in a higher fitness.
        decorrelation_fitness = 1.0 - avg_pairwise_corr
        
        # (Optional) Calculate Sharpe just for logging/analysis purposes
        portfolio_daily_return = selected_returns.mean(axis=1) 
        port_ann_return = portfolio_daily_return.mean() * 252
        port_ann_vol = portfolio_daily_return.std() * np.sqrt(252)
        portfolio_sharpe = port_ann_return / port_ann_vol if port_ann_vol > 0 else 0

        fitnesses = (portfolio_sharpe * 0.5) + (decorrelation_fitness * 0.5) # You can weight these as you see fit

        ant_portfolios.append(selected_indices)
        ant_fitnesses.append(fitnesses)
        ant_portfolios_sharpe.append(portfolio_sharpe)
        
    # Find the best ant in this specific iteration based on DECORRELATION
    iter_best_idx = np.argmax(ant_fitnesses)
    iter_best_portfolio = ant_portfolios[iter_best_idx]
    iter_best_fitness = ant_fitnesses[iter_best_idx]
    iter_best_sharpe = ant_portfolios_sharpe[iter_best_idx]
    
    # Update global best if this iteration found a better decorrelated portfolio
    if iter_best_fitness > global_best_fitness:
        global_best_fitness = iter_best_fitness
        global_best_portfolio = iter_best_portfolio
        global_best_sharpe = iter_best_sharpe
        
    convergence_curve.append(global_best_fitness)
    
    # Pheromone Update Phase
    # 1. Evaporation (Apply to all assets)
    pheromones *= (1 - evaporation_rate)
    
    # 2. Deposit (Reward the assets chosen by the best ant of the iteration)
    for idx in iter_best_portfolio:
        pheromones[idx] += Q * iter_best_fitness

# ---------------------------------------------------------
# Step 5: Output the Results
# ---------------------------------------------------------
best_tickers = [tickers[i] for i in global_best_portfolio]
best_avg_correlation = 1.0 - global_best_fitness # Reverse engineer the fitness to show raw correlation

print("\n--- ACO Optimization Complete ---")
print(f"Best Portfolio Average Correlation: {best_avg_correlation:.4f}")
print(f"Best Portfolio Decorrelation Score (Fitness): {global_best_fitness:.4f}")
print(f"Incidental Sharpe Ratio of this Portfolio: {global_best_sharpe:.4f}")

print("\nTop 30 Selected Assets for Maximum Decorrelation:")
for i, ticker in enumerate(best_tickers, 1):
    print(f"{i}. {ticker}")

# Plot the convergence
plt.figure(figsize=(10, 5))
plt.plot(convergence_curve, marker='o', color='b')
plt.title('ACO Convergence: Portfolio Decorrelation Score over Iterations')
plt.xlabel('Iteration')
plt.ylabel('Decorrelation Score (1 - Avg Correlation)')
plt.grid(True)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch

# ---------------------------------------------------------
# Step 1 & 2: Data Cleaning & Heuristics (Keep as Pandas/NumPy)
# ---------------------------------------------------------
# Assuming 'close_data' is already defined in your environment
threshold = int(0.10 * len(close_data))
clean_data = close_data.dropna(thresh=len(close_data) - threshold, axis=1)
clean_data = clean_data.ffill().bfill()

returns = clean_data.pct_change().dropna()

tickers = clean_data.columns.tolist()
num_assets = len(tickers)
target_assets = 100

full_corr_matrix = returns.corr()
avg_asset_corr = (full_corr_matrix.sum(axis=0) - 1) / (num_assets - 1)
heuristic_info = 1.0 - avg_asset_corr
heuristic_info = heuristic_info - heuristic_info.min() + 0.01

# ---------------------------------------------------------
# Step 3: PyTorch GPU Setup & Initialization
# ---------------------------------------------------------
# Automatically select GPU if available, else fallback to CPU
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Executing on: {device}")

# Move data to PyTorch Tensors on the GPU
returns_tensor = torch.tensor(returns.values, dtype=torch.float32, device=device)
full_corr_tensor = torch.tensor(full_corr_matrix.values, dtype=torch.float32, device=device)
heuristic_tensor = torch.tensor(heuristic_info.values, dtype=torch.float32, device=device)

# ACO Hyperparameters
num_ants = 200
num_iterations = 1000
alpha = 1.0
beta = 2.0
evaporation_rate = 0.1
Q = 1.0

# Pheromones live on the GPU
pheromones = torch.ones(num_assets, dtype=torch.float32, device=device)

global_best_portfolio = None
global_best_fitness = -np.inf
global_best_sharpe = 0
convergence_curve = []

# Pre-calculate constant for annualization
sqrt_252 = torch.tensor(np.sqrt(252), dtype=torch.float32, device=device)

# ---------------------------------------------------------
# Step 4: Run the Vectorized ACO Algorithm
# ---------------------------------------------------------
for iteration in range(num_iterations):
    
    # 1. Calculate probabilities for all assets
    probabilities = (pheromones ** alpha) * (heuristic_tensor ** beta)
    probabilities /= probabilities.sum()
    
    # Expand probabilities so all ants can sample simultaneously
    # Shape: (num_ants, num_assets)
    probs_batch = probabilities.unsqueeze(0).expand(num_ants, -1)
    
    # 2. Vectorized Ant Selection (All 100 ants pick 30 assets at once)
    # Shape: (num_ants, target_assets)
    selected_indices = torch.multinomial(probs_batch, target_assets, replacement=False)
    
    # 3. FAST Decorrelation Fitness Calculation (Batch Matrix Indexing)
    # Extract the pre-calculated correlation sub-matrices for all ants at once
    idx_row = selected_indices.unsqueeze(2) # Shape: (100, 30, 1)
    idx_col = selected_indices.unsqueeze(1) # Shape: (100, 1, 30)
    
    # Resulting shape: (100, 30, 30)
    batch_corr_matrices = full_corr_tensor[idx_row, idx_col] 
    
    # The sum of the off-diagonal elements is the total sum minus the diagonal (which is always 1.0 * target_assets)
    sum_corrs = batch_corr_matrices.sum(dim=(1, 2))
    avg_pairwise_corr = (sum_corrs - target_assets) / (target_assets * (target_assets - 1))
    decorrelation_fitness = 1.0 - avg_pairwise_corr
    
    # 4. FAST Sharpe Calculation (Batch Matrix Multiplication)
    # Create a sparse mask to calculate daily returns for all portfolios at once
    mask = torch.zeros((num_ants, num_assets), device=device)
    mask.scatter_(1, selected_indices, 1.0 / target_assets)
    
    # Returns shape: (Days, Assets). Mask.T shape: (Assets, Ants)
    # Resulting portfolio_returns shape: (Days, Ants)
    portfolio_returns = torch.matmul(returns_tensor, mask.T)
    
    port_ann_return = portfolio_returns.mean(dim=0) * 252
    port_ann_vol = portfolio_returns.std(dim=0, unbiased=True) * sqrt_252
    
    # Avoid division by zero
    portfolio_sharpe = torch.where(port_ann_vol > 0, port_ann_return / port_ann_vol, torch.zeros_like(port_ann_return))
    
    # 5. Combined Fitness
    fitnesses = (portfolio_sharpe * 0.5) + (decorrelation_fitness * 0.5)
    
    # 6. Find Iteration Best
    iter_best_idx = torch.argmax(fitnesses)
    iter_best_portfolio = selected_indices[iter_best_idx]
    iter_best_fitness = fitnesses[iter_best_idx].item()
    iter_best_sharpe = portfolio_sharpe[iter_best_idx].item()
    
    # Update global best
    if iter_best_fitness > global_best_fitness:
        global_best_fitness = iter_best_fitness
        global_best_portfolio = iter_best_portfolio.cpu().numpy() # Pull back to CPU for storage
        global_best_sharpe = iter_best_sharpe
        
    convergence_curve.append(global_best_fitness)
    
    # 7. Pheromone Update Phase
    pheromones *= (1 - evaporation_rate)
    pheromones[iter_best_portfolio] += Q * iter_best_fitness

# ---------------------------------------------------------
# Step 5: Output the Results
# ---------------------------------------------------------
best_tickers = [tickers[i] for i in global_best_portfolio]
best_avg_correlation = 1.0 - global_best_fitness

print("\n--- GPU ACO Optimization Complete ---")
print(f"Best Portfolio Average Correlation: {best_avg_correlation:.4f}")
print(f"Best Portfolio Decorrelation Score (Fitness): {global_best_fitness:.4f}")
print(f"Incidental Sharpe Ratio of this Portfolio: {global_best_sharpe:.4f}")

plt.figure(figsize=(10, 5))
plt.plot(convergence_curve, marker='o', color='b')
plt.title('GPU ACO Convergence: Portfolio Decorrelation Score over Iterations')
plt.xlabel('Iteration')
plt.ylabel('Decorrelation Score (1 - Avg Correlation)')
plt.grid(True)
plt.show()
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# Step 1: Data Cleaning & Log Return Calculation
# ---------------------------------------------------------
# Drop tickers that are missing more than 10% of their data
threshold = int(0.10 * len(close_data))
clean_data = close_data.dropna(thresh=len(close_data) - threshold, axis=1)

# Forward-fill remaining missing data, then backward-fill any leading NaNs
clean_data = clean_data.ffill().bfill()

print(f"Original shape: {close_data.shape}")
print(f"Cleaned shape: {clean_data.shape}")

# Calculate Daily Log Returns
log_returns = np.log(clean_data / clean_data.shift(1)).dropna()

# ---------------------------------------------------------
# Step 2: Calculate Heuristic Information (eta)
# ---------------------------------------------------------
# Calculate annualized metrics using log returns
annualized_log_returns = log_returns.mean() * 252
annualized_log_volatility = log_returns.std() * np.sqrt(252)

# Individual Sharpe Ratio (assuming risk-free rate = 0 for simplicity)
individual_sharpe = annualized_log_returns / annualized_log_volatility

# ACO probabilities must be positive. Shift heuristic values so the minimum is slightly above 0.
heuristic_info = individual_sharpe - individual_sharpe.min() + 0.01

# ---------------------------------------------------------
# Step 3: Ant Colony Optimization Setup
# ---------------------------------------------------------
tickers = clean_data.columns.tolist()
num_assets = len(tickers)
target_assets = 30

# ACO Hyperparameters
num_ants = 30           
num_iterations = 50     
alpha = 1.0             
beta = 2.0              
evaporation_rate = 0.1  
Q = 1.0                 

# Initialize pheromones equally across all assets
pheromones = np.ones(num_assets)

# Tracking the best overall portfolio
global_best_portfolio = None
global_best_fitness = -np.inf
convergence_curve = []

# ---------------------------------------------------------
# Step 4: Run the ACO Algorithm
# ---------------------------------------------------------
for iteration in range(num_iterations):
    ant_portfolios = []
    ant_fitnesses = []
    
    for ant in range(num_ants):
        # Calculate selection probabilities: (Pheromone^alpha) * (Heuristic^beta)
        probabilities = (pheromones ** alpha) * (heuristic_info.values ** beta)
        probabilities /= probabilities.sum() 
        
        # Ant selects exactly 30 unique assets
        selected_indices = np.random.choice(
            range(num_assets), 
            size=target_assets, 
            replace=False, 
            p=probabilities
        )
        
        # Evaluate Fitness: Portfolio Log Return Sharpe Ratio
        selected_log_returns = log_returns.iloc[:, selected_indices]
        
        # For an equally weighted portfolio, the portfolio log return is approximated 
        # by the mean of the individual asset log returns for that day.
        portfolio_daily_log_return = selected_log_returns.mean(axis=1) 
        
        port_ann_return = portfolio_daily_log_return.mean() * 252
        port_ann_vol = portfolio_daily_log_return.std() * np.sqrt(252)
        
        portfolio_sharpe = port_ann_return / port_ann_vol if port_ann_vol > 0 else 0
        
        ant_portfolios.append(selected_indices)
        ant_fitnesses.append(portfolio_sharpe)
        
    # Find the best ant in this specific iteration
    iter_best_idx = np.argmax(ant_fitnesses)
    iter_best_portfolio = ant_portfolios[iter_best_idx]
    iter_best_fitness = ant_fitnesses[iter_best_idx]
    
    # Update global best
    if iter_best_fitness > global_best_fitness:
        global_best_fitness = iter_best_fitness
        global_best_portfolio = iter_best_portfolio
        
    convergence_curve.append(global_best_fitness)
    
    # Pheromone Update Phase
    # 1. Evaporation
    pheromones *= (1 - evaporation_rate)
    
    # 2. Deposit
    for idx in iter_best_portfolio:
        pheromones[idx] += Q * iter_best_fitness

# ---------------------------------------------------------
# Step 5: Output the Results
# ---------------------------------------------------------
best_tickers = [tickers[i] for i in global_best_portfolio]

print("\n--- ACO Optimization Complete (Log Returns) ---")
print(f"Best Portfolio Sharpe Ratio: {global_best_fitness:.4f}")
print("\nTop 30 Selected Assets:")
for i, ticker in enumerate(best_tickers, 1):
    print(f"{i}. {ticker}")

# Plot the convergence
plt.figure(figsize=(10, 5))
plt.plot(convergence_curve, marker='o', color='g')
plt.title('ACO Convergence: Best Portfolio Log-Return Sharpe Ratio')
plt.xlabel('Iteration')
plt.ylabel('Sharpe Ratio')
plt.grid(True)
plt.show()
```

```python
import yfinance as yf
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

# 1. Download data for multiple tickers
tickers = list(all_sector_map.keys())
data = yf.download(tickers, period="1y")['Close']

# 2. Calculate Correlation Matrix
corr = data.pct_change().corr()

# 3. Create Graph object
G = nx.Graph()

# Add nodes
for ticker in tickers:
    G.add_node(ticker)

# Add edges only if correlation is strong (e.g., > 0.7)
threshold = 0.5
for i in range(len(corr.columns)):
    for j in range(i):
        if abs(corr.iloc[i, j]) > threshold:
            G.add_edge(corr.columns[i], corr.columns[j], weight=corr.iloc[i, j])

# 4. Plotting
plt.figure(figsize=(32, 16))
pos = nx.spring_layout(G, k=0.5)  # Position nodes using Fruchterman-Reingold

# Draw nodes, labels, and edges
nx.draw_networkx_nodes(G, pos, node_size=700, node_color='skyblue')
nx.draw_networkx_labels(G, pos, font_size=10, font_family='sans-serif')
nx.draw_networkx_edges(G, pos, width=2, edge_color='gray', alpha=0.5)

plt.title("Stock Correlation Network")
plt.axis('off')
plt.show()
```

```python
import yfinance as yf
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.cm as cm


# 3. Create Graph object
G = nx.Graph()

# Add nodes
for ticker in tickers:
    G.add_node(ticker)

# Add edges only if correlation is strong (e.g., > 0.5)
threshold = 0.2
for i in range(len(corr.columns)):
    for j in range(i):
        correlation_value = corr.iloc[i, j]
        if abs(correlation_value) > threshold:
            # ใช้ค่าสัมบูรณ์ (absolute) เป็น weight เพื่อให้ layout ดึงดูดเข้าหากัน
            G.add_edge(corr.columns[i], corr.columns[j], weight=abs(correlation_value))

# 4. Clustering (Community Detection)
# ใช้อัลกอริทึม Louvain ในการหาคลัสเตอร์ตาม weight ของเส้นเชื่อม
communities = nx.community.louvain_communities(G, weight='weight')

# กำหนดสีให้กับแต่ละคลัสเตอร์
color_map = {}
# สร้างชุดสีตามจำนวนคลัสเตอร์ที่พบ
colors = cm.rainbow([n / (len(communities) - 1) if len(communities) > 1 else 0 for n in range(len(communities))])

for i, community in enumerate(communities):
    for node in community:
        color_map[node] = colors[i]

# นำสีไปใส่ในลิสต์ตามลำดับของโหนดในกราฟ (ถ้าไม่มีกลุ่มให้เป็นสีเทา)
node_colors = [color_map.get(node, 'lightgray') for node in G.nodes()]

# 5. Plotting
plt.figure(figsize=(32, 16))
# Position nodes using Fruchterman-Reingold โดยอ้างอิงจาก weight 
pos = nx.spring_layout(G, k=0.1, weight='weight')  

# Draw nodes, labels, and edges
nx.draw_networkx_nodes(G, pos, node_size=300, node_color=node_colors)
nx.draw_networkx_labels(G, pos, font_size=16, font_family='sans-serif')
nx.draw_networkx_edges(G, pos, width=2, edge_color='gray', alpha=0.5)

plt.title("Stock Correlation Network (Clustered by Correlation)")
plt.axis('off')
plt.show()
```

```python
import numpy as np
import pandas as pd
import yfinance as yf
import networkx as nx
import matplotlib.pyplot as plt
from sklearn.covariance import GraphicalLassoCV
from sklearn.preprocessing import StandardScaler

# 1. Get Data (S&P 500 sectors for a clearer graph)
tickers = list(all_sector_map.keys())

data = yf.download(tickers, period="1y")['Close'].fillna(1)  # Fill NaNs with a small value to avoid issues in correlation

# 2. Preprocessing: Returns and Scaling
returns = data.pct_change().fillna(0)  # Fill NaNs with small value to avoid issues
X = StandardScaler().fit_transform(returns)

# 3. Estimate the GMRF using Graphical Lasso
model = GraphicalLassoCV()
model.fit(X)
precision = model.precision_

# 4. Build the Graph from the Precision Matrix
# We ignore the diagonal (self-correlation)
adj = precision.copy()
np.fill_diagonal(adj, 0)

G = nx.from_numpy_array(adj)
mapping = {i: ticker for i, ticker in enumerate(tickers)}
G = nx.relabel_nodes(G, mapping)

# 5. Visualize
plt.figure(figsize=(10, 8))
pos = nx.spring_layout(G, k=0.3, iterations=50)

# Only draw edges with a significant partial correlation
edges = [(u, v) for u, v, d in G.edges(data=True) if abs(d['weight']) > 0.1]

nx.draw_networkx_nodes(G, pos, node_size=800, node_color='lightgreen')
nx.draw_networkx_labels(G, pos, font_size=9)
nx.draw_networkx_edges(G, pos, edgelist=edges, width=2, edge_color='teal', alpha=0.6)

plt.title("Stock GMRF (Graphical Lasso Sparse Structure)")
plt.axis('off')
plt.show()
```
